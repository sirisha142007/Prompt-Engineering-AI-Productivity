```python

import os
import re
import numpy as np
import faiss
import gradio as gr

from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from transformers import pipeline
```

```python

# Embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# Language model
generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base",
    device=0 if __import__("torch").cuda.is_available() else -1
)

print("AI models loaded successfully!")
```

```python

from google.colab import files

uploaded = files.upload()

pdf_path = next(iter(uploaded.keys()))

print("Uploaded document:", pdf_path)
```

```python

def extract_text(pdf_path):
    reader = PdfReader(pdf_path)

    text = ""

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n"

    return text

document_text = extract_text(pdf_path)

print("Text extracted successfully!")
print(document_text[:1000])
```

```python

def chunk_text(text, chunk_size=500, overlap=100):

    words = text.split()
    chunks = []

    for i in range(0, len(words), chunk_size - overlap):

        chunk = " ".join(words[i:i + chunk_size])

        if chunk.strip():
            chunks.append(chunk)

    return chunks

chunks = chunk_text(document_text)

print("Total chunks:", len(chunks))
print("\nFirst chunk:\n", chunks[0])
```

```python

embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True,
    show_progress_bar=True
)

embeddings = embeddings.astype("float32")

print("Embeddings generated!")
print("Embedding shape:", embeddings.shape)
```

```python

dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

# Normalize embeddings for cosine similarity
faiss.normalize_L2(embeddings)

index.add(embeddings)

print("FAISS vector database created!")
print("Total vectors:", index.ntotal)
```

```python

def retrieve_documents(question, top_k=3):

    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    faiss.normalize_L2(query_embedding)

    scores, indices = index.search(
        query_embedding,
        min(top_k, index.ntotal)
    )

    results = []

    for idx in indices[0]:
        if idx >= 0:
            results.append(chunks[idx])

    return results
```

```python

def ask_question(question):

    if not question.strip():
        return "Please enter a question."

    retrieved_chunks = retrieve_documents(question)

    context = "\n".join(retrieved_chunks)

    prompt = f"""
    Answer the question using only the provided context.

    If the answer is not available in the context,
    say: The information is not available in the document.

    Context:
    {context}

    Question:
    {question}

    Answer:
    """

    response = generator(
        prompt,
        max_new_tokens=200,
        do_sample=False,
        truncation=True
    )

    return response[0]["generated_text"]
```

```python

question = "What is the main topic of this document?"

answer = ask_question(question)

print("Question:", question)
print("\nAI Answer:", answer)
```

```python

with gr.Blocks(title="Chat With Your Documents") as demo:

    gr.Markdown(
        "# 📚 Chat With Your Documents"
    )

    gr.Markdown(
        "Upload a PDF and ask questions using AI-powered RAG."
    )

    question = gr.Textbox(
        label="Enter your question",
        placeholder="Ask something about your document..."
    )

    answer = gr.Textbox(
        label="AI Answer",
        lines=6
    )

    submit = gr.Button("Ask Question")

    submit.click(
        fn=ask_question,
        inputs=question,
        outputs=answer
    )

demo.launch(share=True)
```
