## 2. Fundamentals of Prompt Engineering

Prompt Engineering is the technique of creating effective instructions for AI models to generate desired outputs.

A well-designed prompt improves the relevance, accuracy, clarity, and usefulness of AI-generated responses.

### 2.1 Components of an Effective Prompt

- **Role:** Defines the identity or expertise the AI should adopt.
- **Context:** Provides background information about the task.
- **Instruction:** Clearly explains what the AI needs to do.
- **Input:** Supplies the information required to complete the task.
- **Constraints:** Specifies limitations, rules, length, or formatting requirements.
- **Output Format:** Defines how the response should be presented.

### 2.2 Importance of Prompt Engineering

Prompt Engineering helps users communicate effectively with AI systems. Clear prompts reduce ambiguity, improve output quality, save time, and make AI responses more useful for real-world tasks.

### 2.3 Basic Prompt Structure

A general prompt can be written using the following structure:

Role + Context + Instruction + Input + Constraints + Output Format

### Example

**Weak Prompt:**

Explain Artificial Intelligence.

**Improved Prompt:**

Act as a computer science tutor. Explain Artificial Intelligence to a beginner using simple language, three real-world examples, and a short summary. Keep the response under 150 words.

The improved prompt provides a role, context, clear instruction, and output constraints.

## 3. Zero-Shot and Few-Shot Prompting

Prompting techniques help guide AI models to perform tasks effectively.

### 3.1 Zero-Shot Prompting

Zero-shot prompting involves asking an AI model to perform a task without providing any examples. The model uses its existing knowledge to understand and respond to the instruction.

**Example:**

Prompt: Classify the following review as Positive, Negative, or Neutral: "The product quality is excellent and delivery was fast."

Expected Output: Positive

### 3.2 Few-Shot Prompting

Few-shot prompting provides a small number of examples within the prompt to demonstrate the expected task or response pattern.

**Example:**

Prompt:

Review: "The food was delicious." → Positive

Review: "The service was terrible." → Negative

Review: "The restaurant was okay." → Neutral

Now classify:

Review: "The staff were friendly and helpful."

Expected Output: Positive

### 3.3 Comparison

| Zero-Shot Prompting | Few-Shot Prompting |
|---|---|
| Does not provide examples | Provides a few examples |
| Relies on instructions alone | Learns the expected pattern from examples |
| Simple and quick to use | Useful for specific formats and patterns |
| Suitable for straightforward tasks | Helpful for complex or customized tasks |

### 3.4 Conclusion

Zero-shot prompting is useful for simple tasks, while few-shot prompting can improve consistency by demonstrating the desired response pattern through examples.

## 4. Roles, Context, Instructions, and Constraints

Effective prompt engineering involves structuring prompts with clear roles, context, instructions, and constraints to guide AI models toward accurate and relevant outputs.

### 4.1 Role

A role defines the expertise or identity the AI should adopt while responding.

**Example:** Act as a Python programming tutor.

### 4.2 Context

Context provides background information that helps the AI understand the situation or task.

**Example:** I am a beginner learning Python programming and have basic knowledge of variables and loops.

### 4.3 Instructions

Instructions clearly describe the task that the AI needs to perform.

**Example:** Explain how functions work in Python and provide a simple example.

### 4.4 Constraints

Constraints specify rules or limitations that the AI must follow while generating the response.

**Example:** Use simple language, keep the explanation under 200 words, and avoid advanced programming concepts.

### 4.5 Combined Prompt Example

Act as a Python programming tutor. I am a beginner learning Python. Explain how functions work using a simple example. Use easy language, include a short code snippet, and keep the explanation under 200 words.

### 4.6 Benefits

- Improves response relevance.
- Reduces ambiguity.
- Controls response length and format.
- Helps generate consistent outputs.
- Makes AI responses more useful for specific tasks.

## 5. Structured Outputs Using JSON

Structured outputs allow AI models to generate information in predefined formats such as JSON, tables, and lists.

JSON (JavaScript Object Notation) is a lightweight data format used to store and exchange structured information.

### 5.1 Importance of Structured Outputs

- Organizes AI-generated information.
- Makes outputs easier to read and process.
- Supports automation and integration with applications.
- Improves consistency across responses.
- Helps extract specific information from text.

### 5.2 Example Prompt

Extract the following information from the text and return it in JSON format with the keys "name", "course", and "skills".

Text: "Sirisha is pursuing B.Tech in Artificial Intelligence and Data Science. She has skills in Python, SQL, and Data Analysis."

### 5.3 Expected Output

```json
{
  "name": "Sirisha",
  "course": "B.Tech Artificial Intelligence and Data Science",
  "skills": ["Python", "SQL", "Data Analysis"]
}
```

### 5.4 Applications

Structured outputs are useful in data extraction, report generation, customer information processing, automation, and software development.

## 6. Prompting for Common AI Tasks

Prompt Engineering can be applied to various tasks by providing clear instructions and specifying the expected output.

### 6.1 Summarization

Summarization is the process of reducing lengthy information into shorter, meaningful content while retaining the important points.

**Prompt Example:**

Summarize the following text in five bullet points using simple language.

Text: Artificial Intelligence is transforming industries by automating tasks, analyzing large amounts of data, improving decision-making, and supporting innovation in healthcare, education, business, and technology.

**Expected Output:**

- AI automates repetitive tasks.
- AI analyzes large amounts of data.
- AI supports decision-making.
- AI is used in healthcare and education.
- AI promotes innovation in technology and business.

### 6.2 Classification

Classification involves assigning information to predefined categories based on its characteristics.

**Prompt Example:**

Classify the following customer feedback as Positive, Negative, or Neutral.

Feedback: "The service was excellent and the staff were very helpful."

**Expected Output:** Positive

### 6.3 Content Generation

Content generation involves using AI to create new content such as emails, articles, descriptions, reports, and ideas.

**Prompt Example:**

Write a professional email requesting an internship opportunity in the field of Artificial Intelligence and Data Science. Keep it concise, polite, and professional.

**Expected Output:**

A professional internship inquiry email containing a subject line, greeting, introduction, request, and closing.

### 6.4 Comparison

| Task | Purpose | Example |
|---|---|---|
| Summarization | Shortens information | Summarizing an article |
| Classification | Categorizes information | Categorizing feedback |
| Content Generation | Creates new content | Writing emails or reports |

### 6.5 Conclusion

Clear and task-specific prompts help AI models perform summarization, classification, and content generation more effectively and consistently.

### 7.1 AI-Assisted Coding

### 7.2 AI-Assisted Debugging

Debugging is the process of identifying and correcting errors in a program.

AI can analyze code, identify potential issues, explain errors, and suggest corrected solutions.

**Example of Incorrect Code:**

```python
numbers = [10, 20, 30]
print(numbers[3])
```

**Problem:**

The list contains only three elements. Python indexing starts from 0, so accessing index 3 causes an IndexError.

**Corrected Code:**

```python
numbers = [10, 20, 30]
print(numbers[2])
```

**Output:**

```text
30
```

### 7.3 Benefits of AI-Assisted Coding

- Faster code development.

- Suggestions for code optimization.

## 8. Prompt Evaluation and Testing Reports

Prompt evaluation is the process of testing and analyzing prompts to determine whether they produce accurate, relevant, consistent, and useful outputs.

Testing prompts helps identify weaknesses and improve prompt quality.

### 8.1 Evaluation Criteria

- **Accuracy:** Does the output contain correct information?
- **Relevance:** Does the response address the requested task?
- **Clarity:** Is the output easy to understand?
- **Consistency:** Does the prompt produce similar quality across repeated tests?
- **Instruction Following:** Does the AI follow all the specified requirements?
- **Format Compliance:** Does the output follow the requested structure?

### 8.2 Prompt Testing Example

**Prompt 1 – Basic:**

Write an email requesting an internship.

**Prompt 2 – Improved:**

Act as a professional career assistant. Write a concise email requesting an internship in Artificial Intelligence and Data Science. Include a subject line, formal greeting, brief introduction, internship request, and professional closing.

### 8.3 Testing Report

| Evaluation Criteria | Basic Prompt | Improved Prompt |
|---|---|---|
| Accuracy | Moderate | Good |
| Relevance | General | Specific |
| Clarity | Limited | Clear |
| Instruction Following | Basic | Detailed |
| Format Compliance | Unspecified | Structured |

### 8.4 Observations

The improved prompt provides a role, context, clear instructions, and specific output requirements. This helps produce a more structured, relevant, and professional response.

### 8.5 Conclusion

Prompt evaluation helps users identify ineffective instructions, refine prompts, and develop reusable templates for reliable AI-assisted workflows.

## 9. Reusable Prompt Templates

Reusable prompt templates are predefined instructions that can be adapted for different tasks by replacing placeholders with specific information.

They help save time, maintain consistency, and improve productivity when using AI tools.

### 9.1 Summarization Template

**Template:**

Act as a summarization assistant. Summarize the following content into [NUMBER] bullet points. Use simple language and include only the key information.

Content: [INSERT CONTENT]

### 9.2 Email Writing Template

**Template:**

Act as a professional email-writing assistant. Write an email to [RECIPIENT] regarding [SUBJECT]. Use a [TONE] tone and keep the email [LENGTH]. Include an appropriate greeting and closing.

### 9.3 Resume Assistance Template

**Template:**

Act as a resume-writing assistant. Improve the following resume content for a [JOB ROLE]. Highlight relevant skills, achievements, and experience. Use professional language and maintain clarity.

Resume Content: [INSERT CONTENT]

### 9.4 Coding Assistance Template

**Template:**

Act as an experienced programming tutor. Write Python code to solve [PROBLEM]. Explain the logic step by step, include comments, and provide sample output.

### 9.5 Research Assistance Template

**Template:**

Act as a research assistant. Research the topic [TOPIC]. Provide an introduction, key findings, important concepts, practical applications, and a concise conclusion. Distinguish verified facts from uncertain information and include credible sources.

### 9.6 Benefits of Reusable Templates

- Reduces repetitive prompt writing.
- Saves time and effort.
- Improves consistency.
- Makes prompts easier to customize.
- Supports efficient AI-assisted workflows.

## 10. AI Productivity Toolkit

The AI Productivity Toolkit is a collection of reusable workflows designed to improve productivity using Artificial Intelligence.

It helps users perform common tasks such as summarizing information, drafting emails, improving resumes, assisting with coding, and supporting research.

The toolkit demonstrates how prompt engineering techniques can be organized into practical productivity workflows.

### Toolkit Modules

1. Text Summarization
2. Email Writing Assistant
3. Resume Improvement Assistant
4. Coding Assistant
5. Research Assistant

The following Python implementation provides reusable prompt templates for these productivity tasks.

````python
# AI Productivity Toolkit

def summarizer(text):
    return f"Summarize the following text into 5 key points:\n\n{text}"


def email_assistant(topic, recipient, tone="Professional"):
    return f"""
Write an email.

Recipient: {recipient}
Topic: {topic}
Tone: {tone}

Include a subject, greeting, body, and closing.
"""


def resume_assistant(content, job_role):
    return f"""
Improve the following resume content for the role of {job_role}.
Highlight relevant skills and achievements.

Resume Content:
{content}
"""


def coding_assistant(problem, language="Python"):
    return f"""
Act as a programming assistant.
Write {language} code to solve the following problem.
Explain the logic and provide sample output.

Problem: {problem}
"""


def research_assistant(topic):
    return f"""
Act as a research assistant.
Provide an introduction, key findings, applications,
and conclusion about the following topic.

Topic: {topic}
"""


print("AI Productivity Toolkit Loaded Successfully!")
````

````python
# Testing the AI Productivity Toolkit

print("---- Summarization ----")
print(summarizer("Artificial Intelligence helps automate tasks and improve productivity."))

print("\n---- Email Assistant ----")
print(email_assistant("Internship Application", "HR Manager"))

print("\n---- Resume Assistant ----")
print(resume_assistant("Python, SQL, Data Analysis", "Data Analyst"))

print("\n---- Coding Assistant ----")
print(coding_assistant("Calculate the average of three numbers"))

print("\n---- Research Assistant ----")
print(research_assistant("Generative AI"))
````
