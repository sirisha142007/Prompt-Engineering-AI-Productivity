```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="total_bill", bins=20, color="skyblue")

plt.title("Distribution of Total Bills")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")

plt.show()
```

Most bills fall within the lower-to-middle price range, while fewer customers have higher bill amounts.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="day", y="tip", palette="Set2")

plt.title("Tip Distribution Across Different Days")
plt.xlabel("Day")
plt.ylabel("Tip Amount ($)")

plt.show()
```

Tip amounts vary across different days, with differences in median tips and the spread of tip values.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="total_bill", y="tip", hue="time")

plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip Amount ($)")
plt.legend(title="Time")

plt.show()
```

There is a positive relationship between total bill and tip amount, meaning customers generally tend to give higher tips when their bills are larger.

```python
plt.figure(figsize=(8, 6))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="Blues", fmt=".2f")

plt.title("Correlation Heatmap of Tips Dataset")

plt.show()
```

Total bill and tip show a positive correlation, indicating that tip amounts tend to increase as total bills increase. Group size also shows relationships with bill and tip amounts.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1: Histogram
sns.histplot(data=df, x="total_bill", bins=20, ax=axes[0, 0])
axes[0, 0].set_title("Distribution of Total Bills")
axes[0, 0].set_xlabel("Total Bill ($)")
axes[0, 0].set_ylabel("Frequency")

# Chart 2: Box Plot
sns.boxplot(data=df, x="day", y="tip", hue="day", legend=False, ax=axes[0, 1])
axes[0, 1].set_title("Tip Distribution Across Days")
axes[0, 1].set_xlabel("Day")
axes[0, 1].set_ylabel("Tip Amount ($)")

# Chart 3: Scatter Plot
sns.scatterplot(data=df, x="total_bill", y="tip", hue="time", ax=axes[1, 0])
axes[1, 0].set_title("Total Bill vs Tip Amount")
axes[1, 0].set_xlabel("Total Bill ($)")
axes[1, 0].set_ylabel("Tip Amount ($)")

# Chart 4: Correlation Heatmap
correlation = df.select_dtypes(include="number").corr()
sns.heatmap(correlation, annot=True, cmap="Blues", fmt=".2f", ax=axes[1, 1])
axes[1, 1].set_title("Correlation Heatmap")

# Dashboard title
fig.suptitle("Data Visualization Dashboard - Tips Dataset", fontsize=16)

plt.tight_layout()

# Save the dashboard
plt.savefig("tips_dashboard.png", dpi=150, bbox_inches="tight")

plt.show()
```

## Key Findings

1. **Histogram:** Most total bills fall within the lower-to-middle range, with fewer high-value bills.
2. **Box Plot:** Tip amounts vary across days, showing differences in median values and spread.
3. **Scatter Plot:** Total bill and tip amount show a positive relationship; larger bills generally receive larger tips.
4. **Correlation Heatmap:** Total bill and tip amount have a positive correlation, while group size is also related to bill and tip values.
